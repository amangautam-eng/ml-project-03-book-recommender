```python
# required columns

req =['isbn10','title','authors','categories','thumbnail','description']
dfbook=df_book[req]
```

```python
dfbook.dropna(inplace=True)
```

```python
dfbook.isnull().sum()
```

```python
dfbook
```

```python
def tolist(str):
    return str.split() 
```

```python
dfbook['authors']=dfbook['authors'].apply(tolist)
dfbook['categories']=dfbook['categories'].apply(tolist)
dfbook['description']=dfbook['description'].apply(tolist)
```

```python
dfbook
```

```python
#dfbook['title']=dfbook['title'].apply(lambda x:[i.replace(" ","") for i in x])
dfbook['authors']=dfbook['authors'].apply(lambda x:[i.replace(" ","") for i in x])
dfbook['categories']=dfbook['categories'].apply(lambda x:[i.replace(" ","") for i in x])
dfbook['description']=dfbook['description'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
dfbook
```

```python
dfbook['tags'] = dfbook['authors']+dfbook['categories']+dfbook['description']
```

```python
final_df= dfbook[['isbn10','title','tags']]

final_df['tags']=final_df['tags'].apply(lambda x: " ".join(x))
final_df['tags']=final_df['tags'].apply(lambda x: x.lower())
```

```python
from nltk.stem import  PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[]

    for x in text.split():
        y.append(ps.stem(x))

    return " ".join(y)
```

```python
final_df['tags'].apply(stem)
```

```python
idx=np.arange(0,6247)
final_df.index = idx
```

## Vectorisation

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(final_df['tags'])
voctors=vectors.toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
simlarity=cosine_similarity(vectors)
```

```python
simlarity[0]
```

```python
def recommend(movie):
    bookind=int(final_df[final_df['title']==movie].index[0])

    distance=simlarity[bookind]
    booklist = sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])[1:6]

    for i in booklist:
        print(final_df.iloc[i[0]]['title'])
    


    
```

```python
recommend('I Am that')
```
